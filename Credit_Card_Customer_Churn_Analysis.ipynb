```python
import pandas as pd

df = pd.read_csv("BankChurners.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
```

```python
print("First 5 rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
# Remove unnecessary prediction columns
df = df.drop(columns=[
    "Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_1",
    "Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_2"
])

print("Cleaning completed!")
print("New dataset shape:", df.shape)
```

```python
print("Customer Status:")
print(df["Attrition_Flag"].value_counts())

print("\nCustomer Status (%):")
print(df["Attrition_Flag"].value_counts(normalize=True).mul(100).round(2))
```

```python
print("Gender:")
print(df["Gender"].value_counts())

print("\nEducation Level:")
print(df["Education_Level"].value_counts())

print("\nIncome Category:")
print(df["Income_Category"].value_counts())

print("\nCard Category:")
print(df["Card_Category"].value_counts())

print("\nMarital Status:")
print(df["Marital_Status"].value_counts())
```

```python
print(df.describe())
```

```python
import matplotlib.pyplot as plt

status_counts = df["Attrition_Flag"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(status_counts.index, status_counts.values)

plt.title("Existing vs Attrited Customers")
plt.xlabel("Customer Status")
plt.ylabel("Number of Customers")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

inactive_analysis = pd.crosstab(
    df["Months_Inactive_12_mon"],
    df["Attrition_Flag"],
    normalize="index"
) * 100

inactive_analysis = inactive_analysis.round(2)

print(inactive_analysis)
```

```python
inactive_analysis.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Customer Attrition by Months of Inactivity")
plt.xlabel("Months Inactive in the Last 12 Months")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=0)
plt.legend(title="Customer Status")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df = pd.read_csv("BankChurners.csv")

columns_to_drop = [
    "Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_1",
    "Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_2"
]

df = df.drop(columns=columns_to_drop)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
age_churn = df.groupby("Attrition_Flag")["Customer_Age"].agg(
    ["mean", "median", "min", "max"]
).round(2)

print(age_churn)
```

```python
import matplotlib.pyplot as plt

df.groupby("Attrition_Flag")["Customer_Age"].mean().plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Average Customer Age by Customer Status")
plt.xlabel("Customer Status")
plt.ylabel("Average Age")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
gender_churn = pd.crosstab(
    df["Gender"],
    df["Attrition_Flag"],
    normalize="index"
) * 100

print(gender_churn.round(2))
```

```python
gender_churn.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Customer Status by Gender")
plt.xlabel("Gender")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=0)
plt.legend(title="Customer Status")
plt.tight_layout()
plt.show()
```

```python
education_churn = pd.crosstab(
    df["Education_Level"],
    df["Attrition_Flag"],
    normalize="index"
) * 100

print(education_churn.round(2))
```

```python
education_churn.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Customer Status by Education Level")
plt.xlabel("Education Level")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=45)
plt.legend(title="Customer Status")
plt.tight_layout()
plt.show()
```

```python
contacts_churn = pd.crosstab(
    df["Contacts_Count_12_mon"],
    df["Attrition_Flag"],
    normalize="index"
) * 100

print(contacts_churn.round(2))
```

```python
credit_limit_analysis = df.groupby("Attrition_Flag")["Credit_Limit"].agg(
    ["mean", "median", "min", "max"]
).round(2)

print(credit_limit_analysis)
```

```python
print("Final dataset shape:", df.shape)
print("\nColumns used in analysis:")
print(df.columns.tolist())
```

```python
# Final project summary

total_customers = len(df)

attrited_customers = (df["Attrition_Flag"] == "Attrited Customer").sum()
existing_customers = (df["Attrition_Flag"] == "Existing Customer").sum()

attrition_rate = (attrited_customers / total_customers) * 100

avg_transaction_count = df.groupby("Attrition_Flag")["Total_Trans_Ct"].mean()
avg_transaction_amount = df.groupby("Attrition_Flag")["Total_Trans_Amt"].mean()
avg_utilization = df.groupby("Attrition_Flag")["Avg_Utilization_Ratio"].mean()

print("===== PROJECT SUMMARY =====")
print("Total Customers:", total_customers)
print("Existing Customers:", existing_customers)
print("Attrited Customers:", attrited_customers)
print("Overall Attrition Rate:", round(attrition_rate, 2), "%")

print("\nAverage Transaction Count:")
print(avg_transaction_count.round(2))

print("\nAverage Transaction Amount:")
print(avg_transaction_amount.round(2))

print("\nAverage Utilization Ratio:")
print(avg_utilization.round(3))
```
